```python
from pathlib import Path

data_dir = Path("data")
data_dir.mkdir(exist_ok=True)
```

```python
import boto3
s3 = boto3.client("s3")
```

```python
response = s3.list_objects_v2(
    Bucket="dsan6000-wikipedia",
    Prefix="hourly_parquet/"
)

for obj in response.get("Contents", []):
    print(obj["Key"])
```

```python
for obj in response.get("Contents", []):
    if obj["Key"].endswith(".parquet"):
        file_path = data_dir / obj["Key"].split("/")[-1]
        s3.download_file("dsan6000-wikipedia", obj["Key"], str(file_path))
        print(f"Downloaded {obj['Key']} to {file_path}")
```
